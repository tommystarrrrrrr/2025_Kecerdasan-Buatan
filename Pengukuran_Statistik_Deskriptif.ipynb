# Pengukuran Statistik Deskriptif

## 1.	Central Tendency

#### a. Mendapatkan Nilai Mean

```python
import numpy as np

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.mean(score)
print("Mean:", x)
```

#### b.	Mendapatkan Nilai Median

```python
import numpy as np

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.median(score)
print("Median:", x)
```

#### c.	Mendapatkan Nilai Modus / Mode

```python
from scipy import stats

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = stats.mode(score, keepdims=True)
print("Modus:", x.mode[0])
print("Frekuensi:", x.count[0])
```

## 2.	Spread

#### a.	Mendapatkan Nilai Range

```python
import numpy as np

data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]
minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum
print("Range:", x)
```

#### b.	Mendapatkan Nilai Percentile

```python
import numpy as np

data = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.percentile(data, 90)
print("Percentile ke-90:", x)
```

#### c.	Mendapatkan Nilai Quartile dan Interquartile Range

```python
import numpy as np

data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]
Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

#### d.	Mendapatkan Nilai Variance dan Standart Deviation

```python
import numpy as np

data = [86, 87, 88, 86, 87, 85, 86]
mean = np.mean(data)
var = np.var(data)
std = np.std(data)
print("Mean:", mean)
print("Variance:", var)
print("Standar Deviasi:", std)
```

### 4. Menghitung Standar Deviasi dari Data Berat Badan Mahasiswa Informatika IK-2E

```python
import numpy as np

# Interval dan frekuensi
interval = [(41, 45), (46, 50), (50, 55), (56, 60), (61, 65), (66, 70)]
frekuensi = [2, 3, 6, 4, 5, 2]

# Titik tengah
tengah = [(a + b) / 2 for a, b in interval]

# Rata-rata tertimbang
mean = np.average(tengah, weights=frekuensi)

# Variansi dan standar deviasi
var = np.average((np.array(tengah) - mean) ** 2, weights=frekuensi)
std = np.sqrt(var)

print("Standar deviasi berat badan:", round(std, 2))
```
