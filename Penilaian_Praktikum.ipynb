# KLASIFIKASI

### 1. PENGOLAHAN DATA

```python
import pandas as pd

# Baca data
df = pd.read_csv('data_pasien.csv')

# Informasi awal
print("Informasi Dataset:")
print(df.info())

# 5 data akhir
print("\n5 Data Terakhir:")
print(df.tail())
```

### 2. PEMBERSIHAN DATA

```python
# Periksa missing value
print("Missing Values Awal:")
print(df.isnull().sum())

# Tangani missing value (isi IMT yang kosong dengan median)
if df['IMT'].isnull().any():
    df['IMT'].fillna(df['IMT'].median(), inplace=True)

# Ubah data kategorik (Ya/Tidak) menjadi numerik (1/0)
df['Merokok'] = df['Merokok'].map({'Ya': 1, 'Tidak': 0})
df['Penyakit'] = df['Penyakit'].map({'Ya': 1, 'Tidak': 0})
df['Olahraga'] = df['Olahraga'].map({'Ya': 1, 'Tidak': 0})  # << Tambahan ini tetap di sini

print("\nMissing Values Setelah Penanganan:")
print(df.isnull().sum())

print("\nData Setelah Kategorik diubah (5 data pertama):")
print(df.head())
```

### 3. PEMROSESAN DATA

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Pisahkan fitur dan label
X = df[['Usia', 'IMT', 'TekananDarah', 'Merokok', 'Olahraga']]
y = df['Penyakit']

# Split data latih dan uji (90:10)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42
)

# Normalisasi fitur numerik
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### 4. KLASIFIKASI DENGAN DECISION TREE

```python
# 4. KLASIFIKASI

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Buat dan latih model Decision Tree
model = DecisionTreeClassifier(random_state=0)
model.fit(X_train_scaled, y_train)

# Prediksi data uji
y_pred = model.predict(X_test_scaled)

# Evaluasi akurasi
print("\nAkurasi Model:", accuracy_score(y_test, y_pred))
print("\nLaporan Klasifikasi:")
print(classification_report(y_test, y_pred))
```
