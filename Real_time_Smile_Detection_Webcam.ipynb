# Praktikum E No 1 Real-time Smile Detection with Webcam

## Menggunakan Visual Studio Code

```python
import cv2

# Muat cascade classifier untuk deteksi wajah dan senyum
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_smile.xml')

# Mengaktifkan webcam
cap = cv2.VideoCapture(0)

while True:
    # Membaca frame dari webcam
    ret, frame = cap.read()
    if not ret:
        break

    # Mengubah gambar ke grayscale (hanya satu channel untuk deteksi)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Deteksi wajah
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    # Untuk setiap wajah yang terdeteksi
    for (x, y, w, h) in faces:
        # Gambarkan kotak hijau di sekitar wajah
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)

        # Potong area wajah untuk deteksi senyuman
        roi_gray = gray[y:y+h, x:x+w]

        # Deteksi senyum di dalam area wajah
        smiles = smile_cascade.detectMultiScale(roi_gray, scaleFactor=1.8, minNeighbors=20, minSize=(25, 25))

        # Untuk setiap senyum yang terdeteksi
        for (sx, sy, sw, sh) in smiles:
            # Gambarkan kotak biru di sekitar senyum
            cv2.rectangle(frame, (x+sx, y+sy), (x+sx+sw, y+sy+sh), (255, 0, 0), 2)

    # Tampilkan frame dengan deteksi wajah dan senyum
    cv2.imshow('Face and Smile Detection', frame)

    # Keluar dari loop jika tombol 'q' ditekan
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Tutup webcam dan jendela tampilan
cap.release()
cv2.destroyAllWindows()
```
