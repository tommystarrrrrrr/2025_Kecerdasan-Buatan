## Mengatasi Missing Values dengan Drop Data

### 1. Hapus Data yang kosong (NaN)

```python
df = pd.read_csv('./Data diri.csv')
df.dropna(inplace=True)

df
```

## Mengatasi Missing Values dengan Nilai Baru

### 2. Ganti Data yang kosong (NaN) menjadi :

### <ul><li>Umur = 50 </li><li>Gaji = 25000000</li></ul>

```python
df = pd.read_csv('./Data diri.csv')
df['Umur'].fillna(50, inplace=True)
df['Gaji'].fillna(25000000, inplace=True)
```

```python
df
```

## Mengatasi Missing Values dengan Nilai Mean

## 3. Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Rata-Rata

```python
df = pd.read_csv('./Data diri.csv')
mean_umur = df['Umur'].mean();
mean_gaji = df['Gaji'].mean();

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)
```

```python
df
```

## Mengatasi Missing Values dengan Nilai Median

## 4. Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Median

```python
df = pd.read_csv('./Data diri.csv')
mean_umur = df['Umur'].median();
mean_gaji = df['Gaji'].median();

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)

df
```

## Mengatasi Missing Values dengan Nilai Mode

## 5. Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Modus

```python
df = pd.read_csv('./Data diri.csv')
mean_umur = df['Umur'].mode()[0];
mean_gaji = df['Gaji'].mode()[0];

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)

df
```

## Mengatasi Wrong Data dengan Ubah Data

## 6. Apabila Ada Umur yang diatas 120 maka ubah menjadi 33

```python
df = pd.read_csv('./Data diri.csv')

for x in df.index:
    if df.loc[x, 'Umur'] > 120:
        df.loc[x, 'Umur'] = 33

df
```

# PRAKTIKUM

```python
# Import library
import pandas as pd

# Load dataset
customers = pd.read_csv('customers.csv')
gasstations = pd.read_csv('gasstations.csv')
products = pd.read_csv('products.csv')
transactions = pd.read_csv('transactions_1k.csv')
yearmonth = pd.read_csv('yearmonth.csv')

# Tambahkan kolom waktu
transactions['Date'] = pd.to_datetime(transactions['Date'])
transactions['DateTime'] = pd.to_datetime(transactions['Date'].astype(str) + ' ' + transactions['Time'])
transactions['Hour'] = transactions['DateTime'].dt.hour
transactions['Day'] = transactions['DateTime'].dt.day_name()
```

### 1. list Top 5 Customers yang memiliki nilai transaksi paling banyak

```python
top_customers = (
    transactions.groupby('CustomerID')['Price']
    .sum()
    .sort_values(ascending=False)
    .head(5)
    .reset_index()
)
top_customers.columns = ['CustomerID', 'TotalTransactionValue']
top_customers
```

### 2. list Top 5 Gas stations yang memiliki nilai transaksi paling banyak

```python
top_gasstations = (
    transactions.groupby('GasStationID')['Price']
    .sum()
    .sort_values(ascending=False)
    .head(5)
    .reset_index()
)
top_gasstations.columns = ['GasStationID', 'TotalTransactionValue']
top_gasstations
```

### 3. list Top 5 Jenis produk yang memiliki nilai transaksi paling banyak

```python
product_value = (
    transactions.groupby('ProductID')['Price']
    .sum()
    .sort_values(ascending=False)
    .head(5)
    .reset_index()
)
top_products = product_value.merge(products, on='ProductID')[['ProductID', 'Description', 'Price']]
top_products.rename(columns={'Price': 'TotalTransactionValue'}, inplace=True)
top_products
```

### 4. deskripsi statistik untuk masing masing hari (23,24,25,26) pada dataset transaction_1k

```python
filtered_days = transactions[transactions['Date'].dt.day.isin([23, 24, 25, 26])]
stats_per_day = filtered_days.groupby(filtered_days['Date'].dt.day)['Price'].describe()
stats_per_day
```

### 5. waktu terbaik (hari dan jam) dimana paling banyak user gas station melakukan transaksi

```python
day_hour = transactions.groupby(['Day', 'Hour']).size().reset_index(name='TransactionCount')
best_time = day_hour.sort_values('TransactionCount', ascending=False).head(5)
best_time
```
